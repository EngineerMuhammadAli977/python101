```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
while True:
    print("\nMenu:")
    print("1. Say Hello")
    print("2. Say Bye")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("Hello!")
    elif choice == "2":
        print("Bye!")
    elif choice == "3":
        print("Exiting...")
        break   # Exit the loop
    else:
        print("Invalid input.")
```

```python
while True:
    user_input = input("Enter 'yes' to continue: ")
    if user_input != "yes":
        break
```

```python
correct_password = "admin123"

while True:
    password = input("Enter your password: ")
    if password == correct_password:
        print("Access granted!")
        break
    else:
        print("Wrong password. Try again.")
```
